# Clean sales CSV

Run the cells in order to clean customer names, flag possible duplicates, sort records, and export `customers_sales_cleaned.csv`.

**Requirements:** Python 3 and `pandas` (install with `%pip install pandas` in a separate cell if needed). Put `customers_sales.csv` in the notebook’s working directory or edit `project_dir` below. The export is written to the same directory and replaces an existing file with that name.

The script’s cleaning logic is preserved. Its `main()` body is split into interactive cells, and the script-directory lookup is replaced with `Path.cwd()` because notebooks do not define `__file__`. Outputs have not been generated because the input CSV was not supplied.

## Imports and cleaning rules

```python
from pathlib import Path

import pandas as pd


ORDER_COLUMNS = ["customer_name", "customer_branch", "file_name", "order_quantity"]
CUSTOMER_NAME_CORRECTIONS = {
    "AutoPart AG": "AutoParts AG",
    "Metalform SA": "MetalForm SA",
}
```

## Set input and output paths

```python
project_dir = Path.cwd()
input_path = project_dir / "customers_sales.csv"
output_path = project_dir / "customers_sales_cleaned.csv"

```

## Read and validate the CSV

```python
sales = pd.read_csv(input_path, dtype=str, keep_default_na=False)
missing_columns = [column for column in ORDER_COLUMNS if column not in sales]
if missing_columns:
    raise ValueError(f"Missing columns in {input_path.name}: {', '.join(missing_columns)}")

```

## Clean customer names

```python
sales["customer_name"] = (
    sales["customer_name"].str.strip().replace(CUSTOMER_NAME_CORRECTIONS)
)

```

## Flag possible duplicates

```python
# Flag every row in a matching group because the CSV has no order IDs or dates.
sales["possible_duplicate"] = sales.duplicated(subset=ORDER_COLUMNS, keep=False)


```

## Sort sales records

```python
# Sort alphabetically by company name, then branch
sales = sales.sort_values(by=["customer_name", "customer_branch"], ascending=True)

```

## Export the cleaned CSV

```python
sales.to_csv(output_path, index=False)
print(f"Saved {len(sales)} rows to {output_path.name}.")
print(f"Rows flagged as possible duplicates: {sales['possible_duplicate'].sum()}")
```
