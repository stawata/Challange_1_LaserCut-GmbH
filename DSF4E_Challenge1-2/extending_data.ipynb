### Extending Data Set

Now, we create function containing all the attributes we want to add to the csv file. Then, we create a set with all the attributes. Afterwards, we use the set and .apply to add the attributes. We check the result using display(sales.head()).  

```python
def bounding_info(file_name):
    """Return bounding width, length, cutting length, number of holes, and bounding box area for one DXF file."""

    path = os.path.join(dxf_folder, file_name)

    width, length = get_dxf_bounds(path)

    if width is None or length is None:
        return pd.Series([None, None, None, None, None])

    number_of_holes = get_number_of_holes(path)
    cutting_length = get_cutting_length(path)

    if number_of_holes is None or cutting_length is None:
        return pd.Series([None, None, None, None, None])

    return pd.Series([
        width,
        length,
        cutting_length,
        number_of_holes,
        width * length
    ])


sales[[
    "bounding_box_width",
    "bounding_length",
    "cutting_length",
    "number_of_holes",
    "bounding_box_area"
]] = sales["file_name"].apply(bounding_info)

display(sales.head())
```
